```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load the image
image = cv2.imread('/content/drive/MyDrive/iris/normalabnormal/normal_image_1.jpg')

# Apply median filter to remove noise
median_filtered_image = cv2.medianBlur(image, 5)  # Kernel size is chosen as 5, you can adjust it as needed

# Apply sharpening using High Pass Filter (HPF)
# Convert the image to grayscale
gray_image = cv2.cvtColor(median_filtered_image, cv2.COLOR_BGR2GRAY)

# Apply Gaussian blur to the grayscale image
blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 0)

# Apply Laplacian filter to the blurred image to obtain high-pass filtered image
laplacian = cv2.Laplacian(blurred_image, cv2.CV_64F)

# Add the high-pass filtered image to the original image to enhance sharpness
sharpened_image = np.uint8(gray_image + laplacian)

# Display the original and preprocessed images
cv2_imshow(image)
cv2_imshow(median_filtered_image)
cv2_imshow(sharpened_image)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load the image
image = cv2.imread('/content/drive/MyDrive/iris/normalabnormal/normal_image_1.jpg')

# Convert the image from BGR to RGB format
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Define the threshold range for the sclera color in RGB format
sclera_lower = np.array([150, 100, 50])  # Lower threshold for sclera color (brownish)
sclera_upper = np.array([200, 150, 100])  # Upper threshold for sclera color (brownish)

# Iterate through each pixel of the image
for i in range(image_rgb.shape[0]):
    for j in range(image_rgb.shape[1]):
        # Get the RGB color of the pixel
        pixel_color = image_rgb[i, j]
        # Check if the pixel color falls within the specified threshold range for sclera color
        if (pixel_color >= sclera_lower).all() and (pixel_color <= sclera_upper).all():
            # Set the pixel color to white (255, 255, 255)
            image_rgb[i, j] = [255, 255, 255]

# Display the resulting image
cv2_imshow(image_rgb)
```

```python
import cv2
import numpy as np

# Load the iris image
iris_image = cv2.imread("/content/drive/MyDrive/iris/normalabnormal/normal_image_1.jpg", cv2.IMREAD_GRAYSCALE)

# Define the boundaries for Heart ROI extraction
left_boundary = int(22/32 * iris_image.shape[1])
right_boundary = int(25/32 * iris_image.shape[1])
top_boundary = int(12/32 * iris_image.shape[0])
bottom_boundary = int(27/32 * iris_image.shape[0])

# Extract the Heart ROI
heart_roi = iris_image[top_boundary:bottom_boundary, left_boundary:right_boundary]

# Binarize the Heart ROI
def binarize_image(image, threshold=127):
    _, binary_image = cv2.threshold(image, threshold, 255, cv2.THRESH_BINARY)
    return binary_image

binary_heart_roi = binarize_image(heart_roi)

# Calculate black and white pixel ratio
total_pixels = heart_roi.shape[0] * heart_roi.shape[1]
black_pixels = np.sum(binary_heart_roi == 0)
white_pixels = np.sum(binary_heart_roi == 255)
black_white_ratio = black_pixels / total_pixels, white_pixels / total_pixels

print("Black pixel ratio in Heart ROI:", black_white_ratio[0])
print("White pixel ratio in Heart ROI:", black_white_ratio[1])
```
